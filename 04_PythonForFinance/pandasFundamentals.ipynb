## Data Selection with pn.iloc[]

https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.iloc.html

```python
import pandas as pd
```

```python
data = pd.read_csv('Exercise\Dataset\Lending-company.csv', index_col= 'StringID')
lending_co_data = data.copy()
lending_co_data.head()
```

```python
lending_co_data[1]
```

```python
lending_co_data['Product']
```

```python
# Obtain the information of the first row
lending_co_data.iloc[1]
```

```python
# info of a specific location in the dataframe, (row 1, column 3)
lending_co_data.iloc[1, 3]
```

```python
# Obtain information of all the values in a specific row
lending_co_data.iloc[1,:]
```

```python
# Obtain information of all the values in a specific column
lending_co_data.iloc[:,3]
```

```python
type(lending_co_data.iloc[1,3])
```

```python
type(lending_co_data.iloc[1,:])
```

```python
type(lending_co_data.iloc[:,3])
```

```python
# Access several sets of data
lending_co_data.iloc[[1,3],:]
```

```python
lending_co_data.iloc[:,[1,3]]
```
