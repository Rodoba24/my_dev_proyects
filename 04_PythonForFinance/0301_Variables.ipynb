## Monte Carlo: Black-Schones-Merton

````python
import numpy as np
import pandas as pd
from scipy.stats import norm
````

$$
d_1 = {ln({S\over K})+(r+{stdev^2\over 2})t}\over {s * \sqrt{t}}
$$

$$
d_2 = d_1 - s * \sqrt{t} = {{ln({S\over K})+(r-{stdev^2\over 2})t}\over {s * \sqrt{t}}}
$$


````python
# S - Stock price
# K - strike price
# r - risk-free rate
# stdev - standard deviation
# T - time horizon (years)

def d1 (S, K, r, stdev, T):
    return (np.log(S / K) + (r + stdev ** 2 / 2) * T) / (stdev * np.sqrt(T))

def d2 (S, K, r, stdev, T):
    return (np.log(S / K) + (r - stdev ** 2 / 2) * T) / (stdev * np.sqrt(T))
````

````python
# Cumulative distribution function (cdf)
# Shows how the data accumulates in time
norm.cdf(0)
````

````python
norm.cdf(0.25)
````

````python
norm.cdf(9)
````

$$ C=SN(d_1) - K*e^{-rt}N(d_2)

````python
def BSM(S, K, r, stdev, T):
    return (S * norm.cdf(d1(S,K,r,stdev,T))) - (K * np.exp(-r*T)* norm.cdf(d2(S,K,r,stdev, T)))
````

## Monte Carlo: Euler Discretization - Part I

````python
import numpy as np
import pandas as pd
from scipy.stats import norm
import matplotlib.pyplot as plt

data = pd.read_csv('Exercise\Dataset\PG_2007_2017.csv', index_col='Date')
````

````python
log_retunrs = np.log(1 + data.pct_change())
````

Euler Discretization

$$S_t = S_{t-1} * e^{((r - {1\over 2}*stdev^2)*\delta_t + stdev \sqrt{\delta_t}*Z_t)}

````python
r = 0.025
````

````python
stdev = log_retunrs.std() * 250 ** 0.5
stdev
````

````python
type(stdev)
````

````python
# trasform into numpy array
stdev = stdev.values
stdev
````

````python
T = 1.0
t_intervals = 250
delta_t = T / t_intervals

iterations = 10000
````

````python
Z = np.random.standard_normal((t_intervals + 1, iterations))
S = np.zeros_like(Z)
S0 = data.iloc[-1]
S[0] = S0
````

````python
for t in range (1, t_intervals + 1):
    S[t] = S[t-1] * np.exp((r - 0.5 * stdev ** 2) * delta_t + stdev * delta_t ** 0.5 * Z[t])
````
