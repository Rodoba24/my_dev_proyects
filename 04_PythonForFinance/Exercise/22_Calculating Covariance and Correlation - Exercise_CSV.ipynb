## Calculating Covariance and Correlation

Consider a portfolio composed of *Walmart* and *Facebook*. Do you expect the returns of these companies to show high or low covariance? Or, could you guess what the correlation would be? Will it be closer to 0 or closer to 1? 

Begin by extracting data for Walmart and Facebook from the 1st of January 2014 until today.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
mydata = pd.read_csv('Dataset\Walmart_FB_2014_2017.csv', index_col = 'Date')
mydata
```

Repeat the process we went through in the lecture for these two stocks. How would you explain the difference between their means and their standard deviations?

```python
sec_returns = np.log(mydata / mydata.shift(1))
print (sec_returns.head())
```

```python
sec_returns[['WMT', 'FB']].mean() * 250
```

```python
sec_returns[['WMT', 'FB']].std() * 250 ** 0.5
```

***

## Covariance and Correlation


\begin{eqnarray*}
Covariance Matrix: \  \   
\Sigma = \begin{bmatrix}
        \sigma_{1}^2 \ \sigma_{12} \ \dots \ \sigma_{1I} \\
        \sigma_{21} \ \sigma_{2}^2 \ \dots \ \sigma_{2I} \\
        \vdots \ \vdots \ \ddots \ \vdots \\
        \sigma_{I1} \ \sigma_{I2} \ \dots \ \sigma_{I}^2
    \end{bmatrix}
\end{eqnarray*}

Covariance matrix:

```python
cov_matrix = sec_returns.cov() * 250
cov_matrix
```

__Interpretacion de resultados__

Covarianza de WMT: 0.031544
Covarianza de FB: 0.082786

Covarianza entre WMT y FB: 0.008777

Correlation matrix:

```python
corr_matrix = sec_returns.corr()
corr_matrix
```

__Interpretacion de resultados__

Correlacion de retornos entre WMT y FB: 0.171746
