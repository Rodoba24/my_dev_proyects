## Calculating the Risk of a Security

*Suggested Answers follow (usually there are multiple ways to solve a problem in Python).*

Load the data for Microsoft (‘MSFT’) for the period ‘2000-1-1’ until today. <br />
Assess the daily and the annual risk of ‘MSFT’. Repeat the exercise for Apple for the same timeframe. 

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
mydata = pd.read_csv('Dataset\MSFT_AAPL_2000_2017.csv', index_col = 'Date')
print (mydata.head())

# Se necesita calcluar el retorno logaritmico porque se va a analisar cada accion por separado
sec_returns = np.log(mydata / mydata.shift(1))
print (sec_returns.head())

```

### MSFT

Daily risk:

```python
print ("mean:" , sec_returns['MSFT'].mean())
print ("std:" , sec_returns['MSFT'].std())
```

Annual risk:

```python
print ("mean:" , sec_returns['MSFT'].mean() * 250)
print ("std:" , sec_returns['MSFT'].std() * 250 ** 0.5)
```

### Apple

Daily risk:

```python
print ("mean:" , sec_returns['AAPL'].mean())
print ("std:" , sec_returns['AAPL'].std())
```

Annual risk:

```python
print ("mean:" , sec_returns['AAPL'].mean() * 250)
print ("std:" , sec_returns['AAPL'].std() * 250 ** 0.5)
```

Store the volatilities of the two stocks in an array called “vols”.

```python
vols = np.array([sec_returns['AAPL'].std() * 250 ** 0.5, sec_returns['MSFT'].std() * 250 ** 0.5])
print (vols)
```
