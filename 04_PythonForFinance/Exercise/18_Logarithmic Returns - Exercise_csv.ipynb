## Logarithmic Returns

```python
import numpy as np
import pandas as pd
#from pandas_datareader import data as wb
import matplotlib.pyplot as plt

MSFT = pd.read_csv('Dataset/MSFT_2000_2017.csv', index_col = 'Date')
MSFT
```

### Log Returns

$$
ln(\frac{P_t}{P_{t-1}})
$$

Calculate the Log returns for Microsoft.

```python
MSFT['log_return'] = np.log(MSFT['Adj Close'] / MSFT['Adj Close'].shift(-1))
print (MSFT['log_return'])
```

Plot the results on a graph.

```python
import matplotlib.pyplot as plt

MSFT['log_return'].plot(figsize=(8,5))
plt.show()
```

Estimate the daily and the annual mean of the obtained log returns.

```python
avg_dr = MSFT['log_return'].mean()
print (avg_dr)
```

```python
avg_ar = MSFT['log_return'].mean() * 250
print (avg_ar)
```

Print the result in a presentable form.

```python
print (str(round(avg_ar,4) * 100) + '%')
```
