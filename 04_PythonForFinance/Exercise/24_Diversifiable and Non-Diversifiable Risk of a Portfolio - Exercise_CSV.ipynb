## Diversifiable and Non-Diversifiable Risk of a Portfolio

Import the same dataset we used in the previous lecture – Microsoft and Apple stock. Timeframe – 1st of January 2000 until today. <br />
*Hint: To save time, we have written the code you need!*

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

sec_data = pd.read_csv('Dataset/MSFT_AAPL_2000_2017.csv', index_col='Date')
sec_data
```

Then, calculate the diversifiable and the non-diversifiable risk of a portfolio, composed of these two stocks:

a) with weights 0.5 and 0.5;

```python
weights = np.array([0.5, 0.5])
```

### Calculating Portfolio Variance

Equal weightings scheme:

```python
sec_return = np.log(sec_data / sec_data.shift(1))
```

Portfolio Variance:

```python
cov_matrix = sec_return.cov() * 250

pfolio_var = np.dot(weights.T, np.dot(cov_matrix, weights))
pfolio_var
```

### Calculating Diversifiable and Non-Diversifiable Risk of a Portfolio

Diversifiable Risk:

```python
# Annual variance
msft_var_a = sec_return[['MSFT']].var() * 250

aapl_var_a = sec_return[['AAPL']].var() * 250
```

Or:

```python
# Annual variance
msft_var_a = sec_return['MSFT'].var() * 250

aapl_var_a = sec_return['AAPL'].var() * 250
```

Calculating Diversifiable Risk:

```python
# diversifiable risk = portfolio variance - weighted annual variances

msft_wav = weights[0] ** 2 * msft_var_a
aapl_wav = weights[1] ** 2 * aapl_var_a

dr = pfolio_var - msft_wav - aapl_wav

print ( round(dr*100,4), '%' )
```

Calculating Non-Diversifiable Risk:

```python
n_dr_1 = pfolio_var - dr
n_dr_1
```

```python
n_dr_2 = msft_wav + aapl_wav
n_dr_2
```
