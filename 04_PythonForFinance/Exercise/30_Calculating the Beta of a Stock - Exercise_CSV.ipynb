## Calculating the Beta of a Stock

*Suggested Answers follow (usually there are multiple ways to solve a problem in Python).*

Load the data for Microsoft and S&P 500 for the period 1st of January 2012 – 31st of December 2016. 

```python
import numpy as np
import pandas as pd

pf_data = pd.read_csv('Dataset\CAPM_Exercise_Data.csv', index_col='Date')
```

```python
pf_data.head()
```

Let S&P 500 act as the market. 

*****

Calculate the beta of Microsoft.

```python
sec_returns = np.log( pf_data / pf_data.shift(1))

# Calculating the covariance matix
cov = sec_returns.cov() * 250
```

```python
cov
```

```python
# Obtain the covariance between the stock and the market
cov_with_market = cov.iloc[0,1]
cov_with_market
```

```python
market_var = sec_returns['^GSPC'].var() * 250
market_var
```

**Beta:** 
$$ 
\beta_{pg} = \frac{\sigma_{pg,m}}{\sigma_{m}^2}
$$

```python
# Calculate Beta:

MSFT_beta = cov_with_market / market_var
MSFT_beta
```
